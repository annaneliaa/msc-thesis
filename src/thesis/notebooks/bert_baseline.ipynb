#### Metrics table

```python
def collect_bert_results(results_root: str | Path) -> pd.DataFrame:
    """
    Collect all scenario-level BERT results from result files into one table.

    Expected structure:
        results_root/
            scenario_1/results.txt
            scenario_2/results.txt
            ...

    Each results.txt should contain lines like:
        accuracy: 0.9943
        precision: 1.0000
        recall: 0.8033
        f1: 0.8909
        ....

    Returns:
        DataFrame with columns:
        [scenario, accuracy, precision, recall, f1, tp, fp, tn, fn, pred_attack_count, pred_attack_rate, avg_attack_prob, max_attack_prob, min_attack_prob]
    """
    results_root = Path(results_root)
    rows = []

    for results_file in results_root.rglob("results.txt"):
        metrics = {}
        scenario = results_file.parent.name

        with open(results_file, "r") as f:
            for line in f:
                line = line.strip()
                if not line or ":" not in line:
                    continue

                key, value = line.split(":", 1)
                key = key.strip()
                value = value.strip()

                try:
                    metrics[key] = float(value)
                except ValueError:
                    metrics[key] = value

        row = {"scenario": scenario, **metrics}
        rows.append(row)

    df = pd.DataFrame(rows)

    metric_cols = ["accuracy", "precision", "recall", "f1", "tp", "fp", "tn", "fn", "pred_attack_count", "pred_attack_rate", "avg_attack_prob", "max_attack_prob", "min_attack_prob"]
    existing_cols = ["scenario"] + [c for c in metric_cols if c in df.columns]
    df = df[existing_cols].sort_values("scenario").reset_index(drop=True)

    return df
```

```python
df = collect_bert_results("../../../out/bert/alerts_csv")
df.to_csv("../../../out/bert/alerts_csv/bert_results_summary.csv", index=False)
```
